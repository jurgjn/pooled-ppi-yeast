```bash
# Start container and store container id
CONTAINER_ID="$(docker run -p 8501:8501 --detach jurgjn/pooled-ppi-yeast)"
```

```bash
# Copy table with pools metadata from container to current directory
docker container cp $CONTAINER_ID:/data/predictions/alphafold3_summary_confidences.parquet .
```

```bash
# Stop container
docker stop $CONTAINER_ID
```

```bash
# Current directory now has a copy of the file
ls -l alphafold3_summary_confidences.parquet
```
